```python
from tensorflow.keras.preprocessing.text import one_hot
```

```python
### sentences
sent=['the glass of milk',
'the glass of juice',
'the cup of tea',
'I am a good boy',
'I am a good developer',
'understand the meaning of words',
'your videos are good',]

sent
```

```python
# Define the vocablary size
voc_size=10000
```

```python
# One Hot Representation for every word
one_hot_repr = [one_hot(words,voc_size) for words in sent]
one_hot_repr
```

```python
# Word Embedding Representation

from tensorflow.keras.layers import Embedding
from tensorflow.keras.layers import Input, Embedding
#from tensorflow.keras.preprocessing.sequence import pad_sequences
from tensorflow.keras.utils import pad_sequences
from tensorflow.keras.models import Sequential


import numpy as np 
```

```python
# To sent the sentences to the model they should be of same length or fix length hence use pad_sequences

sent_length = 8 # max sentence length
embedded_docs = pad_sequences(one_hot_repr,padding='pre',maxlen=sent_length)
# To make the sentences of same length the pad_sequence will add 0s in the end(post) or in the start(pre)
print(embedded_docs)
```

```python
# Feature Represenation

dim = 10
```

```python
# Embedding Layer internally using Word2Vec, No training Needed

model = Sequential()
model.add(Input(shape=(sent_length,)))
model.add(Embedding(voc_size, dim))
model.compile('adam', 'mse')
```

```python
model.summary()
```

```python
# every word will be represented as 10 dimension

model.predict(embedded_docs)
```
